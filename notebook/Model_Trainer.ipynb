```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error,r2_score
from sklearn.model_selection import RandomizedSearchCV
from sklearn.model_selection import KFold
```

```python
df=pd.read_csv('data/cleaned_data.csv')
df.head()
```

```python
x=df.drop('Item_Outlet_Sales',axis=1)
y=df['Item_Outlet_Sales']
```

```python
x.head()
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_train_scaled
```

```python
x_test_scaled=scaler.transform(x_test)
x_test_scaled
```

#### **Model Training**

##### 1. Linear Regression

```python
from sklearn.linear_model import LinearRegression
linear_model=LinearRegression()
linear_model.fit(x_train_scaled,y_train)
```

```python
print("Training Score: ",linear_model.score(x_train_scaled,y_train))
```

```python
y_pred=linear_model.predict(x_test_scaled)
```

```python
rmse=root_mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
print(f"Root Mean Squared Error: {rmse}")
print(f"R2 Score: {r2}")
```

#### 2. Lasso Regression

```python
cv=KFold(n_splits=5)
param_grid={
    'alpha':[0.01,0.1,1,10,100],
    'max_iter':[1000,2000,3000,4000],
    'selection':['cyclic','random']
}
```

```python
from sklearn.linear_model import Lasso
lasso_model=Lasso()
lasso_random=RandomizedSearchCV(estimator=lasso_model,param_distributions=param_grid,cv=cv,n_iter=10,scoring='r2',n_jobs=-1,verbose=2,random_state=42)
lasso_random.fit(x_train_scaled,y_train)
```

```python
print("Training Score: ",lasso_random.best_score_)
```

```python
lasso_random.best_params_
```

```python
lasso_y_pred=lasso_random.predict(x_test_scaled)
```

```python
print(f"Root Mean Squared Error: {root_mean_squared_error(y_test,lasso_y_pred)}")
print(f"R2 Score: {r2_score(y_test,lasso_y_pred)}")
```

#### 3. Ridge Regression

```python
param_grid1={
    'alpha':[0.01,0.1,1,10,100],
    'max_iter':[1000,2000,3000,4000],
    'solver':['auto', 'svd', 'cholesky', 'lsqr', 'sparse_cg', 'sag', 'saga', 'lbfgs']
}
```

```python
from sklearn.linear_model import Ridge
ridge_model=Ridge()
ridge_random=RandomizedSearchCV(estimator=ridge_model,param_distributions=param_grid1,cv=cv,n_iter=10,scoring='r2',n_jobs=-1,verbose=2,random_state=42)
ridge_random.fit(x_train_scaled,y_train)
```

```python
print("Training Score: ",ridge_random.best_score_)
ridge_y_pred=ridge_random.predict(x_test_scaled)
print(f"Root Mean Squared Error: {root_mean_squared_error(y_test,ridge_y_pred)}")
print(f"R2 Score: {r2_score(y_test,ridge_y_pred)}")
```

#### 4. Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
rf_model=RandomForestRegressor()
param_grid2={
    'n_estimators':[100,200,300,400,500],
    'max_depth':[6,7,8,9,10],
    'min_samples_split':[8,9,10,11],
    'min_samples_leaf':[20,22,24,25]
}
rf_random=RandomizedSearchCV(estimator=rf_model,param_distributions=param_grid2,cv=cv,n_iter=100,n_jobs=-1,verbose=2,random_state=42,scoring='r2')
rf_random.fit(x_train_scaled,y_train)
```
